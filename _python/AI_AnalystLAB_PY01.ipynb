## 1. The IDLE editor — where we write real programs

In Session 0 we typed single lines at the `>>>` prompt. That prompt — the **shell** — is lovely for one quick line, like scribbling a single note on a sticky pad. But real programs have **many lines**, and the shell takes only one at a time. For anything longer we use the **editor**: think of it as a clean **page** where you write your whole recipe, then run it all at once.

Here is how, step by step:

1. In IDLE, click the **File** menu, then **New File**. A second, empty window opens — that is your editor page.
2. Type (or paste) your code onto the page.
3. Save it: **File → Save**, name it something like **`meet_python.py`**, and click Save. *(The `.py` ending simply tells the computer "this is a Python file.")*
4. Run it: from the menu at the top, click **Run → Run Module**. *(There is also a keyboard shortcut — **F5** — but please read the Mac note just below before you reach for it.)*
5. The result appears back in the **shell** window (the one with `>>>`).

> **🍎 Mac users — did pressing F5 pop up "Start Dictation?"** On most modern Macs the **F5** key is the *Dictation / microphone* key, so tapping it opens a **"Start Dictation?"** prompt instead of running your code. That is not an error — just click **Cancel** (or press the **Esc** key). The reliable way to run your program on a Mac is the menu: **Run → Run Module**. *(If you like keyboard shortcuts, hold **fn** and press **F5** together.)*

> **⚠️ A few more things that trip up everyone at first — all completely normal:**
> - If you try to run *before* saving, IDLE pops up a box asking you to save. Just click **OK / Save** and it runs.
> - Make sure you are typing on the **editor page**, not back in the `>>>` shell. If a chunk of code refuses to run, that is the usual reason.
> - We will build **one single file** today. Each time we learn something new, **add it to the bottom of the same page and run it again** (Run → Run Module). You will see all the results together — that is exactly what we want.

## 2. A program is just instructions, in order

A **program** is a list of instructions the computer follows **from top to bottom**, one after another — exactly like following a recipe step by step. The instruction we already met, `print(...)`, means *"show this on the screen."*

Type these two lines onto your editor page and run it (menu: **Run → Run Module**):

````python
print("Meet a passenger aboard the Titanic.")
print("We will describe this one person, step by step.")
````

The two sentences appear in the shell, in the order you wrote them. Simple — but that top-to-bottom order *is* the whole idea of a program.

***
## 3. Variables — labelled boxes for your data

We don't want to *retype* our passenger's details every time. Instead we store them in **variables**. A variable is a **labelled box**: you give it a name, put a value inside, and later you just say the name to get the value back.

You make one with an **equals sign**:

```
age = 22
```

Read that as *"put 22 into a box labelled `age`,"* **not** as "age equals 22" from mathematics. The `=` is an **action** — *put into* — not a statement of fact.

Let's give our passenger five details. Add these to your file and run it (menu: **Run → Run Module**):

````python
name = "Braund, Mr. Owen Harris"
age = 22
sex = "male"
survived = False
fare = 7.25

print(name)
print(age)
````

The shell shows the name and the age. The other three boxes (`sex`, `survived`, `fare`) are filled too — we just haven't printed them yet.

> **⚠️ Two easy slips:**
> - Variable names are **case-sensitive**: `age` and `Age` are *two different boxes*. Pick one spelling and keep to it.
> - If you ever see a red **`NameError: name 'age' is not defined`**, it almost always means a **typo** in the name, or that the line *making* the box comes *after* the line *using* it. Instructions run top to bottom — so make the box first.

***
## 4. Four everyday kinds of value

Look again at those five boxes. Their contents are not all the same *kind* of thing — just as, in a kitchen, *"3 eggs,"* *"the word* flour *on a jar,"* and *"yes, we have milk"* are different kinds of information. Python has four kinds you will use constantly:

| Kind | Called | Example from our passenger |
|---|---|---|
| a whole number | **int** (integer) | `age = 22` |
| a decimal number | **float** | `fare = 7.25` |
| a piece of text | **str** (string) | `name = "Braund, Mr. Owen Harris"` |
| a yes / no value | **bool** (boolean) | `survived = False` |

Text always goes inside **quotation marks**. A **bool** is only ever `True` or `False` (capital first letter, no quotes). You can ask Python what kind a value is with **`type(...)`**. Add this and run it (menu: **Run → Run Module**):

````python
print(type(age))       # a whole number
print(type(fare))      # a decimal number
print(type(name))      # text
print(type(survived))  # yes/no
````

Python answers `<class 'int'>`, `<class 'float'>`, `<class 'str'>`, `<class 'bool'>` — its own names for the four kinds. You don't need to memorise this. Just remember: **every value has a kind**, and Python is **fussy about mixing kinds** — which is exactly what the next section shows.

***
## 5. Making readable sentences — f-strings

We usually want to combine words *and* our stored values into a sentence, like *"Owen was 22 years old."* Here is where beginners hit a very common wall. You might try to glue text and a number together with `+`:

```
print("Age: " + age)
```

Python **refuses** this, with a red `TypeError`, because `"Age: "` is text and `age` is a number, and it will not silently mix the two kinds *(remember: Python is fussy about kinds)*. **That error is not you doing something wrong — it is Python being careful.**

The clean, modern fix is an **f-string**: put the letter **`f`** just before the opening quotation mark, and drop any variable inside **curly braces `{ }`**. Python fills in the blanks — like a fill-in-the-blank sentence. Add this and run it (menu: **Run → Run Module**):

````python
print(f"{name} was {age} years old and paid {fare} pounds for the ticket.")
print(f"Did this passenger survive? {survived}")
````

Out comes a proper sentence with the values slotted into place. **This is the single most-used tool in the whole AI Analyst LAB course** — every result you ever report is printed with an f-string. It is well worth getting comfortable with it now.

***
## 6. A little arithmetic

Python is also a calculator. The everyday operators are just what you'd expect:

`+` add · `-` subtract · `*` multiply · `/` divide · `**` "to the power of"

Suppose the passenger had bought a *second* identical ticket. Add this and run it (menu: **Run → Run Module**):

````python
two_tickets = fare * 2
print(f"Two tickets would cost {two_tickets} pounds.")
print(f"Half a ticket would cost {fare / 2} pounds.")
````

Notice `fare / 2` gives a **decimal** (a float). Dividing with `/` always gives a decimal answer in Python — even when the numbers divide evenly. A small thing, but worth knowing.

***
## 7. Asking yes/no questions — comparisons

Very often we need the computer to *judge* something: is this passenger an adult? A **comparison** asks such a yes/no question and answers with a **bool** (`True` or `False`):

`==` equal to · `!=` not equal to · `<` less than · `>` greater than · `<=` / `>=`

Add this and run it (menu: **Run → Run Module**):

````python
print(f"Is the passenger an adult (18 or older)? {age >= 18}")
print(f"Is the passenger female? {sex == 'female'}")
````

The first answer is `True` (22 is 18 or more); the second is `False` (our passenger's `sex` is `"male"`). These true/false answers are the seed of **decisions** (coming in Session 3) and, later, of picking particular rows out of a data table.

> **⚠️ The classic mix-up:** `=` (one equals sign) *puts a value into a box*; `==` (two equals signs) *asks whether two things are equal*. Writing `=` where you meant `==` is the most common beginner slip of all — when a comparison behaves strangely, check this first.

***
## 8. Comments — notes to your future self

Any line that starts with a **`#`** is a **comment**: Python ignores it completely. Comments are notes *for humans* — like scribbling in the margin of a cookbook. Use them to remind yourself what a piece of code is for. Add this and run it (menu: **Run → Run Module**):

````python
# This whole line is just a note — Python skips it entirely.
total = fare + 5   # add a 5-pound booking fee (a note at the end of a real line of code)
print(f"With a 5-pound booking fee, the total is {total} pounds.")
````

Good comments make code readable again weeks later. You will see them throughout this course.
