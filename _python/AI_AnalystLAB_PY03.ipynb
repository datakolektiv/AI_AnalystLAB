## 1. A quick refresher — True and False

In Session 1 you saw that a **comparison** answers a yes/no question with a `bool` — `True` or `False`:

````python
print(22 >= 18)            # is 22 at least 18?
print("male" == "female")  # are these two pieces of text the same?
````

These `True` / `False` answers are the fuel for **decisions** — which is exactly where we start.

## 2. Making decisions — `if`, `elif`, `else`

A decision lets your program take a different path depending on whether something is true — like a **fork in the road**, or a bouncer who checks an ID before deciding whether to let someone in. The shape is:

```
if (some yes/no question):
    do this
else:
    do that instead
```

Two small details do all the work here, and both catch beginners:

- The line ends with a **colon `:`**.
- The lines that belong *inside* the `if` are **indented** — pushed to the right by four spaces. Python uses that indentation, and nothing else, to know which lines are inside.

Add this and run it:

````python
age = 22

if age >= 18:
    print("This passenger is an adult.")
else:
    print("This passenger is a child.")
````

> **⚠️ The two most common beginner errors live right here:**
> - **Forgetting the colon** at the end of the `if` (or `else`) line → a red **`SyntaxError`**.
> - **Getting the indentation wrong** → a red **`IndentationError`**. The indented lines must line up. Good news: after you type the `:` and press Enter, IDLE indents the next line *for* you.

For **more than two** paths, add `elif` (short for *"else, if…"*) in the middle — as many as you like:

````python
age = 70

if age < 18:
    print("Child")
elif age < 65:
    print("Adult")
else:
    print("Senior")
````

Python checks the questions from top to bottom and runs the **first** one that is true, then skips the rest.

***
## 3. Doing something for every item — the `for` loop

So far we have handled one value at a time. A **loop** repeats a piece of code for *every* item in a collection — like a **conveyor belt** bringing each item past you in turn, or dealing a deck of cards one by one.

The shape:

```
for each_item in the_collection:
    do something with each_item
```

`each_item` is a name **you** choose; on each pass through the loop it holds the next value. Let's loop over a small list of ages:

````python
ages = [22, 38, 26, 35, 35]

for age in ages:
    print("Age:", age)
````

The indented line ran once for each value — five times. (Same indentation rule as `if`: the loop's body is the indented part.)

***
## 4. Looping over our passengers

Now the real thing. Our `passengers` is a list of dictionaries, so a loop hands us **one passenger (one dictionary) at a time**. We reach inside each one with its labels, exactly as in Session 2:

````python
for passenger in passengers:
    print(passenger["name"], "— survived:", passenger["survived"])
````

And by putting an `if` *inside* the loop, we can act only on certain passengers — for instance, print just the survivors:

````python
for passenger in passengers:
    if passenger["survived"]:
        print("Survived:", passenger["name"])
````

We can even loop over **one passenger's own labels** using `.keys()` (from Session 2) to visit every field of a record:

````python
first_passenger = passengers[0]

for label in first_passenger.keys():
    print(label, "→", first_passenger[label])
````

That is the promise from Session 2 kept: `.keys()` hands us the labels, and we look up each value in turn.

***
## 5. Counting and totalling — the accumulator

To answer *"how many survived?"* we use the **accumulator** pattern: start a counter at zero, then add to it as the loop runs — like clicking a **tally counter** once for each survivor. The `+=` means *"add this to whatever is already there."*

````python
survivors = 0                       # start the tally at zero
for passenger in passengers:
    if passenger["survived"]:
        survivors += 1              # one more survivor: add 1

print("Number of survivors:", survivors)
````

The same idea totals numbers up. To find the **average age**, we sum all the ages, then divide by how many there are:

````python
total_age = 0
for passenger in passengers:
    total_age += passenger["age"]

average_age = total_age / len(passengers)
print("Average age:", average_age)
````

You just computed two real summaries — a count and an average — from raw data, with nothing but a loop, an `if`, and a running total. That is the beating heart of data analysis.

***
## 6. A shortcut — list comprehensions

Very often we loop for one specific reason: to build a **new list** out of an old one. That is so common that Python has a compact one-line shortcut for it, called a **list comprehension**. Let's see the long way first, then the short way.

**The long way** — make an empty list, then append to it in a loop:

````python
ages_long = []
for passenger in passengers:
    ages_long.append(passenger["age"])
print(ages_long)
````

**The short way** — a *list comprehension* does exactly the same thing in a single line. Read it left to right as a sentence: *"the age of each passenger, for every passenger in passengers."*

````python
ages_short = [passenger["age"] for passenger in passengers]
print(ages_short)
````

Same result, one tidy line. The shape is always `[ something  for each_item in the_collection ]`.

This little pattern is worth getting comfortable with, because **it is everywhere in the AI Analyst LAB course.** Every simulation you will run — like drawing ten thousand random averages to watch a bell curve appear — is a list comprehension: `[ ... for _ in range(10000) ]`. Once you recognise the shape, that code stops looking mysterious.

## 7. Functions — write it once, use it again

Suppose you want the survival rate not just now, but for many different groups of passengers. Rather than copy-and-paste the loop each time, you package it into a **function**: a named recipe you write **once** and can "make" whenever you like, by giving it an **input** and getting a **result** back. Think of a coffee machine — you press the button and out comes coffee; you do not rebuild the machine every morning.

You create one with **`def`** (short for *define*), and hand back the result with **`return`**:

````python
def survival_rate(people):
    survivors = 0
    for person in people:
        if person["survived"]:
            survivors += 1
    return survivors / len(people)     # hand the answer back to whoever called
````

Defining a function does not *run* it — it just teaches Python the recipe. To actually use it, you **call** it by name, passing in some data. The value it `return`s comes back to you, ready to use:

````python
rate = survival_rate(passengers)
print("Survival rate:", rate)
print("That is", round(rate * 100), "percent of our passengers.")   # round() gives a whole number
````

> **⚠️ `return` is not `print`.** `print` only *shows* a value on the screen, and then it is gone. `return` **hands the value back** so you can store it (here, in `rate`) and keep using it — as we just did to work out a percentage. A function that only prints cannot be reused in a calculation; a function that returns can. When in doubt, **`return`** the answer.

The whole AI Analyst LAB course is built from small functions exactly like this — in the reviews you will write and use ones such as `entropy()` and `information_gain()`. A function is simply how you give a useful idea a name.

## 8. When Python turns red — reading an error calmly

Sooner or later a line will turn **red**. That is not Python scolding you — it is Python **trying to help**, pointing at the exact spot that confused it. The trick is knowing where to look.

An error message (a *traceback*) can look long, but the important part is almost always **the very last line**. It names the *kind* of error and usually says, in plain-ish English, what went wrong — and just above it, it shows the line that tripped.

Here is a field guide to the ones you have already met:

| The last line says… | It usually means… |
|---|---|
| **`SyntaxError`** | a typo in the *shape* of the code — often a missing `:` , quote, or bracket |
| **`IndentationError`** | the spaces at the start of a line do not line up |
| **`NameError`** | you used a name Python has not seen — a typo, or a box you never made |
| **`TypeError`** | you mixed kinds that don't go together (like text `+` a number) |
| **`IndexError`** | a list position that doesn't exist (positions start at 0) |
| **`KeyError`** | a dictionary label that isn't there — check the spelling |

**The routine:** read the last line, fix the thing it names on the line it points to, and run again. And — as always — **if you are stuck for more than a few minutes, don't wrestle with it: contact your course instructor.** A fresh pair of eyes fixes most errors in seconds.
