## 2. Lists — an ordered row of values

A **list** is the simplest container: an **ordered row of values**, like a numbered shopping list or a row of lockers. You write it inside **square brackets `[ ]`**, with commas between the values.

Here are the ages of our first five passengers. Add this to `passengers.py` and run it (**Run → Run Module**):

````python
ages = [22, 38, 26, 35, 35]

print(ages)
print("How many ages are in the list?", len(ages))
````

`len(ages)` tells us the list holds 5 values. Now — how do we get *one* value out? By its **position**, written in square brackets. Here is the single most important quirk for a beginner to absorb: **Python counts positions from 0, not 1.** So the *first* value is at position `0`, the *second* at `1`, and the *fifth* at `4`. Add this and run it:

````python
print("The first age (position 0):", ages[0])
print("The second age (position 1):", ages[1])
print("The last age (position -1):", ages[-1])
````

Notice `ages[-1]` — a **negative** position counts from the *end*, so `-1` is the last value. Handy.

> **⚠️ The number-one beginner error: `IndexError`.** Our list has five values, at positions **0, 1, 2, 3, 4** — there is no position `5`. Asking for one that doesn't exist:
> ```
> print(ages[5])
> ```
> gives a red **`IndexError: list index out of range`**. If you see it, remember: the last position is *one less* than the count (or just use `-1`).

You can also take a **slice** — a range of positions — using a colon. `ages[0:3]` means *"from position 0 up to (but not including) 3"*, i.e. the first three:

````python
print("The first three ages:", ages[0:3])
````

Unlike a tuple (next section), a list can be **changed**. You can replace a value, or **add** one with `.append()`. Suppose one passenger has a birthday, and a new passenger boards:

````python
ages[0] = 23            # Owen just had a birthday: change position 0
ages.append(40)         # a new passenger, aged 40, boards: add to the end
print("Updated ages:", ages)
````

Finally — and this matters later — a list does **not** care whether its values are all the same kind. It will happily hold a passenger's *mixed* details — text, a number, and a true/false — side by side:

````python
first_passenger = ["Braund, Mr. Owen Harris", 22, "male", False]
print(first_passenger)
````

Text, a number, and `False` living together in one list, no complaint. **Remember that** — it will matter a great deal in Sessions 4 and 5.

## 3. Tuples — a fixed record

A **tuple** looks almost like a list, but it is written with **round brackets `( )`** — and it has one crucial difference: **once made, it cannot be changed.** Think of it as a stamped ID card or a sealed record: you can *read* what is on it, but not alter it.

A tuple is the natural way to hold **one record made of related-but-different fields** — like a single passenger's *(name, age, sex, class, fare)*:

````python
owen_tuple = ("Braund, Mr. Owen Harris", 22, "male", 3, 7.25)

print(owen_tuple)
print("Name:", owen_tuple[0])   # read a field by position, just like a list
print("Age: ", owen_tuple[1])
````

There is a lovely shortcut for opening a record into named boxes all at once, called **unpacking** — list the names on the left, the tuple on the right:

````python
p_name, p_age, p_sex, p_class, p_fare = owen_tuple
print(f"{p_name} was {p_age} years old, travelling in class {p_class}.")
````

> **⚠️ A tuple cannot be changed.** Trying to overwrite a field:
> ```
> owen_tuple[1] = 23
> ```
> gives a red **`TypeError: 'tuple' object does not support item assignment`**. That is not a bug — it is the whole *point* of a tuple: a fixed record you can trust not to change by accident. When you *do* need to edit, use a **list** instead.

**Rule of thumb:** a **list** for many of the *same* kind of thing (all the ages); a **tuple** for one fixed record of *different* fields (one passenger).

## 4. Dictionaries — look things up by label, not position

With the tuple above, was `owen_tuple[3]` the class or the fare? You had to *remember* the order. A **dictionary** fixes this by giving every value a **label** (called a **key**). You then look things up by name, not by counting — like finding *"phone"* in a contacts entry instead of remembering it is the fourth line.

A dictionary uses **curly brackets `{ }`**, and each entry is written **`label: value`**:

````python
owen_dict = {
    "name": "Braund, Mr. Owen Harris",
    "age": 22,
    "sex": "male",
    "survived": False,
    "fare": 7.25,
}

print("Age:      ", owen_dict["age"])       # look up by label — clear and readable
print("Survived: ", owen_dict["survived"])
````

`owen_dict["age"]` is so much clearer than `owen_tuple[1]` — you say what you want *by name*.

> **⚠️ Ask for a label that isn't there and you get a `KeyError`.**
> ```
> print(owen_dict["cabin"])
> ```
> gives **`KeyError: 'cabin'`** — there is no `"cabin"` entry. Check the spelling of the key, or add it first.

You can also **add** a brand-new entry, or **change** an existing one, exactly like editing a list value:

````python
owen_dict["age"] = 23                  # change an existing entry
owen_dict["embarked"] = "Southampton"  # add a brand-new entry
print(owen_dict)
````

#### Seeing all the labels, or all the values — `.keys()` and `.values()`

Sometimes you want the **labels** on their own, or the **values** on their own, rather than looking a single one up. A dictionary gives you two small tools for exactly this. Picture the dictionary as a **form with labelled boxes**:

- **`.keys()`** hands you *just the labels* — the questions printed on the form (`name`, `age`, `sex`, …).
- **`.values()`** hands you *just the answers* — what is written in the boxes, in the same order as the labels.

The `.` before each one means *"a thing a dictionary can do."* We wrap each in `list( )` simply so it prints as a tidy list. Add this and run it:

````python
print("The labels (keys):   ", list(owen_dict.keys()))
print("The answers (values):", list(owen_dict.values()))
````

So `owen_dict.keys()` gives back **every label** —

`['name', 'age', 'sex', 'survived', 'fare', 'embarked']`

— and `owen_dict.values()` gives back **every value**, lined up in the same order —

`['Braund, Mr. Owen Harris', 23, 'male', False, 7.25, 'Southampton']`

This is handy when you want to know *what fields a record has* (its keys) without caring about the contents, or to scan *all the contents at once* (its values). In **Session 3** you will use exactly these to walk through every field of a passenger in a loop.

#### A closer look at the dot — *actions* vs. *facts*

Look again at what you just typed: `owen_dict.keys()` and `owen_dict.values()`. That shape — a **thing**, a **dot** `.`, a name, and a pair of **parentheses `( )`** — turns up *everywhere* in Python, so let's slow right down and name it.

The **dot** `.` means *"something that belongs to this thing."* And what belongs to a thing comes in two flavours — and **the parentheses tell you which**:

- **An action the thing can *do*** — its **method**. You make it happen by adding **`( )`**, which mean *"do it now."* `owen_dict.keys()` is a method: you are telling the dictionary, *"go and hand me your labels."* `owen_dict.values()` is another.
- **A fact the thing simply *has*** — its **attribute**. You read it with **no parentheses** at all — just the name.

An everyday picture. Think of a **car** as a "thing":

- `car.colour` would be an **attribute** — a *fact the car has* (say, `"red"`). No parentheses; you just read it off.
- `car.start()` would be a **method** — an *action the car does*. The `( )` are you pressing the button.

So here is the one test to keep in your head whenever you see a dot: **is there a `( )` after it?** With `( )` → it is an *action* (a method). Without `( )` → it is a *value the thing holds* (an attribute).

The dictionaries we have met so far only offer **methods** — `.keys()` and `.values()` are actions you *run*. Clear **attributes** turn up a little later: in **Sessions 4–5** a table of data will have a `.shape` — a plain fact, *how many rows and columns it has* — that you read with **no** parentheses (`table.shape`). You will recognise it the instant you see it.

> **⚠️ The mix-up this saves you from.** Forget the `( )` on a method and you do *not* get an error — you get something odd. Writing `print(owen_dict.keys)` *without* the parentheses prints a strange line like `<built-in method keys of dict object …>` instead of the labels. If you ever see output like that, you almost certainly left the `( )` off.

*(One thing we are doing on purpose: we keep calling these **"things."** In **Session 8** you will meet the proper word — every one of them is an **object**, and *methods* and *attributes* are simply the actions and facts that belong to it. For now, "thing" is all you need.)*

Dictionaries are the workhorse of data work: the labels are self-documenting, so code stays readable. This exact shape — labelled fields — is how the AI Analyst LAB course later describes data to Claude's API.

## 5. Putting it together — a list of dictionaries is a tiny table

Now the payoff. If **one passenger** is a dictionary, then **many passengers** is simply a **list of dictionaries** — and that is exactly the shape of a table: each dictionary is a **row**, each label is a **column**.

Here are our five passengers together. Add this and run it:

````python
passengers = [
    {"name": "Braund, Mr. Owen Harris",     "age": 22, "sex": "male",   "survived": False},
    {"name": "Cumings, Mrs. John Bradley",   "age": 38, "sex": "female", "survived": True},
    {"name": "Heikkinen, Miss. Laina",       "age": 26, "sex": "female", "survived": True},
    {"name": "Futrelle, Mrs. Jacques Heath", "age": 35, "sex": "female", "survived": True},
    {"name": "Allen, Mr. William Henry",     "age": 35, "sex": "male",   "survived": False},
]

print("We have", len(passengers), "passengers.")
print("The first passenger:", passengers[0])
print("The first passenger's name:", passengers[0]["name"])
````

Read that last line carefully: `passengers[0]` picks the **first row** (a dictionary), and `["name"]` reaches into it for the **name field**. *Position* to choose the row, *label* to choose the field.

This little list-of-dictionaries is the seed of everything that follows. In **Session 5**, the **pandas** toolbox will turn a structure exactly like this into a real, powerful **DataFrame** — a spreadsheet inside Python — with a single command.

## 6. The one idea to carry forward

Look back at what we just did. Every passenger record mixed **text** (`name`, `sex`), a **number** (`age`), and a **true/false** (`survived`) — and lists, tuples, and dictionaries all held that mixture **without complaint**. Python's everyday containers are wonderfully flexible about the *kinds* of value inside them.

**Hold onto this thought — it is one of the most important in the whole crash course:**

- In **Session 4** you meet **NumPy**, whose *arrays* are extremely fast at maths but **refuse to mix kinds** — put text and numbers together and NumPy quietly turns *everything* into text.
- In **Session 5** you meet **pandas** *DataFrames*, which mix kinds the *smart* way: **one kind per column**, many columns per table.

Knowing which container mixes types, and which does not, is a big part of what makes a data analyst effective. You have just met the flexible three; the stricter, faster tools come next.
