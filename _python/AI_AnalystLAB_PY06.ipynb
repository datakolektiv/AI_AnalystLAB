## 2. The question that filtering can't answer neatly

Last session you kept **one** group at a time — *the survivors* — with a mask. But real questions compare **many** groups at once: *how did survival differ between **women and men**? Between **first, second, and third class**?*

You *could* do it by hand: filter the women, compute their rate; then filter the men, compute theirs; then write both down. Tedious — and it gets worse the more groups there are. pandas has a single tool that does **all the groups in one stroke**. But first, a gentle warm-up.

***
## 3. A quick headcount — `.value_counts()`

Before summarising, analysts often just ask *"how many of each?"* — a plain **headcount** of a column. That's **`.value_counts()`**: hand it a column, and it tallies how many times each value appears — like sorting a bag of coloured beads into piles and counting each pile.

```python
print(passengers["Sex"].value_counts())
```

Out come the tallies: **577 male, 314 female**. Try it on ticket class too:

```python
print(passengers["Pclass"].value_counts())
```

**491** passengers in 3rd class, **216** in 1st, **184** in 2nd.

> Notice the order: `.value_counts()` lists the **most common first** (3rd class, then 1st, then 2nd) — not `1, 2, 3`. It sorts by *how many*, not by the value itself. Good to know so the order never surprises you.

***
## 4. The main event — `.groupby()`

Here is the analyst's workhorse. Picture a teacher with a big pile of exam papers: they **split** the papers into stacks — one stack per class — then work out the **average score for each stack**, and finally **combine** those averages into a short list. **`.groupby()`** does exactly that, in one line: *split* the table into groups, *summarise* each group, *combine* the answers into a small table.

Let's ask the real question — the survival **rate** for each sex:

```python
survival_by_sex = passengers.groupby("Sex")["Survived"].mean()
print(survival_by_sex.round(3))
```

Read the line left to right, like a sentence:

- **`passengers.groupby("Sex")`** — split the table into groups by the `Sex` column (one group of women, one of men);
- **`["Survived"]`** — in each group, look at the `Survived` column;
- **`.mean()`** — take its average within each group;
- **`.round(3)`** — tidy the answer to 3 decimals so it's easy to read.

The result: **women `0.742`, men `0.189`** — about **74%** of the women survived, against only **19%** of the men. In one line, the Titanic's grim *"women and children first"* is written in the data.

> **Why is an *average* a *rate*?** The `Survived` column is all **1**s (survived) and **0**s (did not). If you average a pile of 1s and 0s, the answer *is* just the **fraction that are 1** — so the **mean of `Survived` is the survival rate**. That small trick — averaging a yes/no column to get a proportion — is used constantly in the main course.

***
## 5. Grouping by two things at once

You can split by **more than one column** — just hand `groupby` a **list** of column names inside `[ ]`. Let's see survival by **sex *and* class** together:

```python
print(passengers.groupby(["Sex", "Pclass"])["Survived"].mean().round(3))
```

Now the story deepens. Read down the six groups:

- **First-class women: `0.968`** — almost everyone survived.
- **Third-class men: `0.135`** — almost no one did.

Sex mattered enormously, but so did class — and the two together tell a sharper story than either alone. That is the whole point of grouping: **let the data split itself and compare.**

*(The `["Sex", "Pclass"]` is a **list** of two columns — the same square-bracket list you met in Session 2 — telling `groupby` to make a group for every sex-and-class combination.)*

***
## 6. Ordering the answer — `.sort_values()`

A summary is easier to read when it's **sorted**. **`.sort_values()`** puts the numbers in order — smallest to largest by default, or largest first if you ask. Let's rank the classes by survival, highest first:

```python
survival_by_class = passengers.groupby("Pclass")["Survived"].mean()
print(survival_by_class.sort_values(ascending=False).round(3))
```

Top to bottom: **1st class `0.630`, 2nd `0.473`, 3rd `0.242`** — survival fell steadily as you moved down the classes. The **`ascending=False`** part simply means *"biggest first"*; leave it out and you get smallest first.

Sorting doesn't change the numbers — it just arranges them so the story reads in order, which matters even more once you start drawing charts of them.
