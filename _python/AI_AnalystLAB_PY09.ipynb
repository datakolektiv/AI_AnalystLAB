## 1. The crack we're here to seal

Remember exactly where Session 8 left off. Our hand-written `Passenger` class took **anything** you gave it:

```python
ghost = Passenger("A. Ghost", "banana", "unknown", 5)   # an age of "banana"... and Python shrugged.
```

No error. The class stored the garbage, and the crash came **later and far away**, when something finally tried to do maths on `"banana"`. We ended with a wish: *what if the blueprint checked each fact the moment the object was born?*

**That wish is pydantic.** Same idea as a class — a blueprint with named fields — but with a guard on the door. Let's build one.

***
## 2. A new kind of blueprint

Open **IDLE → File → New File**, save it (into your `AI_AnalystLAB` folder) as **`validated.py`**. As with last session, there's **nothing to download** — pydantic is a toolbox you installed back in Session 0.

First we bring it in, then we write our blueprint — and it looks a little different from last session's class:

````python
from pydantic import BaseModel, ValidationError

class Passenger(BaseModel):
    name: str
    age: int
    sex: str
    survived: int
````

Compare this to the class you wrote by hand last session — it's actually **less** typing:

- **`class Passenger(BaseModel):`** — the `(BaseModel)` part means *"build this on top of pydantic's ready-made blueprint."* It's how you borrow all of pydantic's checking powers. *(You'll meet the deeper idea behind "building on top of" — inheritance — properly in the main course; here just read it as "give my blueprint pydantic's superpowers.")*
- **`name: str`**, **`age: int`**, … — you simply **name each field and state its kind**: `name` is text (`str`), `age` is a whole number (`int`). That little `: str` is not a comment — it is a **rule** pydantic will enforce.
- **No `__init__`, no `self` lines!** Because you declared the fields, **pydantic writes the set-up recipe for you.** All that boilerplate from last session — gone.

> **⚠️ If `from pydantic import …` turns red** with **`ModuleNotFoundError`**, pydantic wasn't installed — redo the Session 0 install, or ask your instructor.

Run the file — again, nothing visible yet. You've drawn the blueprint; now let's use it.

***
## 3. Building a passenger — and a lovely surprise

Create Owen from the blueprint. One important difference from last session: with pydantic you **name each field** as you pass it — `age=22`, not just `22`:

````python
owen = Passenger(name="Braund, Mr. Owen Harris", age=22, sex="male", survived=0)

print(owen)
````

Look at what `print(owen)` gives you now:

`name='Braund, Mr. Owen Harris' age=22 sex='male' survived=0`

Every field, laid out plainly — a world away from the cryptic `<…Passenger object at 0x…>` you got last session. That's the "prints beautifully" gift we promised: pydantic makes your objects **readable for free**. And you still reach a single fact the same familiar way, with a dot and no parentheses:

````python
print("Name:", owen.name)
print("Age: ", owen.age)
````

> **⚠️ Name your fields.** With pydantic you must pass fields **by name** — `Passenger(name="…", age=22, …)`. Handing them over bare and in order (as you could last session) won't work here. Naming them is clearer anyway — you can see exactly what each value is for.

***
## 4. The magic — it *checks*

Now the whole point. Let's hand pydantic the same nonsense that a plain class swallowed — an age of `"banana"` — and watch what happens.

To do it safely, we'll meet one small **new** tool: **`try` / `except`**. It is Python's way of saying *"**try** to run this; if it fails, don't crash the whole program — jump to the **`except`** part and handle it calmly instead."* Back in Session 3 you learned to *read* Python's red error messages; `try` / `except` is how you *catch* one on purpose. Here we use it so pydantic's complaint prints as a tidy line rather than a scary crash:

````python
try:
    ghost = Passenger(name="A. Ghost", age="banana", sex="unknown", survived=5)
except ValidationError as error:
    print("pydantic refused to build it. Here's exactly why:\n")
    print(error)
````

There it is. Instead of silently storing junk, pydantic **stopped you at the door** and told you precisely what was wrong and where:

> *age — Input should be a valid integer, unable to parse string as an integer …*

The mistake is caught **the instant the object is created** — not hours later in some far-off calculation. That is the entire gift of pydantic: **bad data can't get in.**

And it's **smart**, not merely fussy. Give it a value it can *sensibly* read as the right kind — the text `"22"` — and it quietly converts it into the number `22`:

````python
tidy = Passenger(name="Neat Example", age="22", sex="male", survived=0)

print("age is now:", tidy.age, "— a real number, converted for us")
````

So `"22"` (text that clearly means a number) is accepted and turned into `22`; `"banana"` (text that means nothing numeric) is refused. Sensible checking — exactly what you'd want a careful assistant to do.

## 5. Full circle — and the bridge to the real course

Cast your mind all the way back to **Session 2**, where Owen first appeared as a plain **dictionary**:

````python
owen_dict = {"name": "Braund, Mr. Owen Harris", "age": 22, "sex": "male", "survived": 0}

owen = Passenger(**owen_dict)     # hand the whole dictionary to the blueprint
print(owen)
````

In one line, a loose, unchecked dictionary became a **guaranteed, validated `Passenger`**. *(The `**owen_dict` is just a tidy way of saying "use each dictionary entry as a named field" — no need to retype them.)* And you can go the other way just as easily — turn a model back into a plain dictionary:

````python
print(owen.model_dump())
````

Out comes `{'name': 'Braund, Mr. Owen Harris', 'age': 22, 'sex': 'male', 'survived': 0}` — a dictionary again, the very shape you started with in Session 2. **Dictionary → validated object → dictionary.** The circle closes.

**And here is why this is the finale.** That flow — *loose data in, checked against a strict shape, tidy data out* — is precisely how modern tools talk to an **AI**. In the AI Analyst LAB course you'll hand a pydantic blueprint like this one to Claude and say, in effect, *"answer me, and your answer **must** fit this shape."* Pydantic guarantees what comes back is clean and usable — no `"banana"` ages slipping through.
