### First, five words explained (no jargon)

Before we install anything, here are the only five words you need — each with a picture from everyday life.

- **Python** — a *language for giving a computer instructions*. Think of it as a very precise **recipe language**: you write the steps, and the computer follows them exactly, in order.

- **IDLE** — the **workspace** that comes *free* with Python. Picture a **notepad with a "Run" button**: you write your recipe on the page, press a key, and watch the computer cook it. (The name stands for *Integrated Development and **Learning** Environment* — the "Learning" part is the whole point; it was made for beginners like you.)

- **A package** (also called a *library*) — a **ready-made toolbox** that someone else built and shared, so you don't start from nothing. Doing data analysis without the `pandas` toolbox would be like **whisking cream by hand when a mixer sits on the counter**. We will install five toolboxes.

- **`pip`** — the **delivery service** for toolboxes. You place one order, and it fetches them from the internet and installs them for you.

- **The Terminal** — a window where you **type commands to the computer directly**, instead of clicking buttons. It looks bare — just a blinking cursor on a plain background — and it can feel a little intimidating the first time. Don't worry: today you will only **paste a single line** into it. Think of it as **ordering at a counter by saying what you want**, rather than pointing at pictures on a menu. You will use it more later in the course, so a first friendly hello now is helpful.

That's the whole vocabulary. Let's begin.

***
## Step 1 — Install Python (this also gives you IDLE)

Installing Python once gives you **three things at the same time**: the Python language, the IDLE workspace, and the `pip` delivery service. Follow **only** the section for your computer.

### 🍎 macOS

1. Open this page in your web browser: **[https://www.python.org/downloads/](https://www.python.org/downloads/)**
2. Click the big yellow button that says **"Download Python 3.x.x"** (the page detects your Mac automatically).
3. When it finishes, open your **Downloads** folder and double-click the file that ends in **`.pkg`**.
4. An installer window opens. Click **Continue → Continue → Continue → Agree → Install**. Enter your **Mac login password** if it asks, then wait for it to finish and click **Close**.
5. **One extra click that prevents a common headache.** After installing, a Finder window usually pops open showing the "Python 3.x" folder. If you see a file called **`Install Certificates.command`**, **double-click it**. A small black window opens, runs by itself for a few seconds, and then you can close it. *(This lets Python download things from the internet safely — which `pip` needs in Step 4.)*

### 🪟 Windows

1. Open this page in your web browser: **[https://www.python.org/downloads/](https://www.python.org/downloads/)**
2. Click the big yellow button that says **"Download Python 3.x.x"**.
3. Open your **Downloads** folder and double-click the file you just downloaded (it ends in `.exe`).
4. **This next part matters a lot.** On the very first installer screen, look at the **bottom** and **tick the checkbox that says "Add python.exe to PATH."** *(This is what lets the Terminal find Python in Step 4. It is easy to miss — please check it.)*
5. Now click **"Install Now"**. Wait for it to finish, then click **Close**.

### 🐧 Linux (Ubuntu / Debian)

Linux usually already has Python, but often *not* IDLE or `pip`. We install all three with one command. First open the Terminal (**Ctrl + Alt + T**), then paste this line and press **Enter**:

```
sudo apt update && sudo apt install -y python3 python3-pip idle3
```

It will ask for your password — the same one you use to log into your computer. *(As you type the password nothing appears on screen; that is normal — just type it and press Enter.)*

***
## Step 2 — Open IDLE

Now let's open the workspace where we will write code.

- **🍎 macOS:** press **Cmd + Space** (this opens Spotlight search), type **`IDLE`**, and press **Enter**. *(Or: Finder → Applications → the "Python 3.x" folder → IDLE.)*
- **🪟 Windows:** click the **Start** menu, type **`IDLE`**, and click **"IDLE (Python 3.x)"**.
- **🐧 Linux:** open your applications menu and search for **`IDLE`**, or type **`idle3`** in the Terminal and press **Enter**.

A window opens with a few lines of text, ending in a line that begins with **`>>>`**.

That `>>>` is the **IDLE shell** — it is the computer politely waiting for you to type an instruction, a bit like a friendly *"Yes? What would you like?"*. We will type here in Steps 5 and 6. **Leave this window open** and move on.

***
## Step 3 — Meet the Terminal

To install our five toolboxes we use the `pip` delivery service, and `pip` is given its orders from the **Terminal**. Here is how to open it. *(Linux users already did this in Step 1 — same window.)*

- **🍎 macOS:** press **Cmd + Space**, type **`Terminal`**, and press **Enter**.
- **🪟 Windows:** click **Start**, type **`PowerShell`**, and click **"Windows PowerShell"**.
- **🐧 Linux:** press **Ctrl + Alt + T**.

A window opens with a blinking cursor and not much else. That is completely normal — it is just waiting. You do **not** need to understand it today; in the next step you will paste **one** line and press Enter.

***
## Step 4 — Install the five toolkits

In the **Terminal window** (from Step 3), copy the line for **your** system, paste it in, and press **Enter**.

- **🍎 macOS**
```
python3 -m pip install numpy pandas matplotlib seaborn pydantic
```

- **🪟 Windows**
```
py -m pip install numpy pandas matplotlib seaborn pydantic
```

- **🐧 Linux (Ubuntu / Debian)**
```
python3 -m pip install --user numpy pandas matplotlib seaborn pydantic
```

> **⚠️ Important — have you used Python on this computer before?** These steps assume Python is **brand new** to your machine. If you have already installed Python at some point and done some earlier work with it (for example through Homebrew, Anaconda, or another tool), your old and new setups can **clash** and produce confusing error messages. **If any command in this notebook gives you an error you did not expect, please do not try to fix it yourself — contact your course instructor right away.** They will get you unstuck in a few minutes. This is exactly the kind of thing that is quick for an instructor to sort out and frustrating to guess at alone.

When you press Enter, a lot of text will scroll by for a minute or two as each toolbox is downloaded and installed. **That scrolling is a good sign** — it means it is working. When it stops and the blinking cursor returns, you are done.

**What did we just order?** The five toolboxes, and what each one is for:

| Toolbox | What it does, in one line |
|---|---|
| **numpy** | fast maths on lots of numbers at once |
| **pandas** | spreadsheets inside Python (tables of data) |
| **matplotlib** | draws charts |
| **seaborn** | draws prettier charts with less effort |
| **pydantic** | checks that data has the shape you expect |

***
## Step 5 — Check that everything arrived

Now go back to the **IDLE window** (the one with `>>>`). Click just to the right of the `>>>`, **type or paste the single line below**, and press **Enter**.

> **How IDLE works — and one small rule.** Whatever you type after `>>>` and press Enter, the computer does right away, showing the answer just underneath. But the `>>>` shell handles **one line at a time** — if you paste *several* lines into it at once, IDLE gets confused and shows `SyntaxError: multiple statements found while compiling a single statement`. That is why today's check is written as a **single line**. *(Starting in **Session 1** you will use IDLE's **editor** — a page where you can write and run many lines together — but for today, one line is all we need.)*

````python
import numpy, pandas, matplotlib, seaborn, pydantic; print("All five toolkits imported successfully — you are ready!")
````

If you see the message **"All five toolkits imported successfully — you are ready!"**, everything is installed. 🎉

If instead you see a red message containing **`ModuleNotFoundError`**, it means one of the five toolboxes did not install. Go back to **Step 4**, run the install command again, then type this line once more.

***
## Step 6 — Say "Hello World"

There is a tradition, as old as programming itself: the first thing you make a computer do is say hello. Let's honour it. In the **IDLE** window, next to `>>>`, type this and press **Enter**:

````python
print("Hello World")
````

The computer answers:

```
Hello World
```

**That's it — you just wrote and ran your first program.** 🎉

What actually happened, in plain words: **`print(...)`** is an instruction that means *"show this on the screen,"* and the words inside the quotation marks are what you asked it to show. You gave the computer an instruction in the Python recipe language, and it obeyed. Everything else in this course is built from small, understandable steps exactly like this one.
