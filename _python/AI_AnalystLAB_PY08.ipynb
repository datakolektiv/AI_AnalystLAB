## 1. The big reveal — you've used objects all along

Here is the secret: **in Python, almost everything is an *object*.** An object is simply a *thing that bundles together two kinds of stuff*:

- **facts about itself** — called **attributes** (you read them with **no** parentheses);
- **actions it can do** — called **methods** (you call them **with** parentheses).

And that **dot** you've been typing? It is how you *talk to* an object — *"hey, table, give me your `.shape`"* (a fact) or *"hey, table, `.head()` yourself"* (an action). You have been doing this since the very beginning:

- `passengers.head()` — `passengers` is an object (a **DataFrame**); `.head()` is one of its actions.
- `ages.mean()` — `ages` is an object (a **column**); `.mean()` is an action.
- even `"hello".upper()` — a plain piece of **text** is an object too, with an `.upper()` action.

So *objects* are not a strange new thing. They are the thing you have been using all along — we are just finally giving them their name, and learning to **make our own**.

***
## 2. Blueprint vs. building — *class* and *object*

To make our own objects we need two words, and one honest analogy.

Think of a **cookie cutter**. The cutter itself is not a cookie — it is the *shape*, the **template** that says "a cookie has this outline." From one cutter you can stamp out **many** cookies, all the same shape, each a separate cookie you can decorate differently.

Programming has exactly this pair:

- A **class** is the cookie cutter — the **blueprint**. It describes *what every object of this kind will look like*: what facts it holds, what actions it can do. **You write the class once.**
- An **object** (also called an **instance**) is a **cookie** — an actual thing stamped out from the blueprint. From one class you make **as many objects as you like**.

So today we will **write one class** — the blueprint for a passenger — and then **stamp out passengers** from it. Let's set up and do it.

***
## 3. Writing your first blueprint — the `Passenger` class

Open **IDLE → File → New File**, and **save it** (into your `AI_AnalystLAB` folder is fine) as **`objects.py`**. *(Good news: there is **nothing to download** today — we are building our own objects from scratch, so no data file is needed.)*

Now type this. Read the explanation underneath **before** you worry about the pieces:

```python
class Passenger:
    def __init__(self, name, age, sex, survived):
        self.name = name
        self.age = age
        self.sex = sex
        self.survived = survived
```

Let's take it apart, line by line — slowly, because this is the densest idea in the whole crash course:

- **`class Passenger:`** — "I am defining a new blueprint, and its name is `Passenger`." (By tradition, class names start with a Capital letter.)
- **`def __init__(self, ...):`** — this is the **set-up recipe**, run automatically every time a new passenger is stamped out. Its odd name, `__init__`, is short for *"initialise"* — *"get this new object ready."* You don't call it yourself; Python runs it for you.
- **`self`** — this means **"this particular passenger being made right now."** When we later stamp out Owen, inside `__init__` `self` *is* Owen; when we stamp out Laina, `self` *is* Laina.
- **`self.name = name`** — "take the `name` I was handed, and **store it on this passenger** as its `name` fact." Same for age, sex, and survived. This is how the object *remembers* its own facts.

> **⚠️ Three things beginners trip on here — check these first if you get a red error:**
> - **The colons `:`** at the end of the `class` line and the `def` line are required.
> - **Indentation is everything.** `def __init__` is indented **inside** the class; the four `self.… = …` lines are indented **again, inside** `__init__`. Wrong indentation gives an **`IndentationError`**. Keep it tidy — four spaces per step.
> - **Don't forget `self`.** Every `self.something` needs it. If your stored facts "vanish" later, a missing `self` is the usual reason. If stuck, contact your instructor.

Run the file now. Nothing visible happens — and that's correct! You've only **drawn the blueprint**; you haven't stamped out any passengers yet.

***
## 4. Stamping out a passenger — and reading its facts

Now the fun part. Below your class, create an actual passenger by **calling the class like a function**, handing it the four facts. Let's build Owen — yes, *the* Owen, passenger `0` from Session 5:

```python
owen = Passenger("Braund, Mr. Owen Harris", 22, "male", 0)

print(owen.name)
print(owen.age)
```

That's it — `owen` is now a real **`Passenger` object**, stamped from your blueprint. Two things to notice:

- To **make** the object you wrote `Passenger("Braund, Mr. Owen Harris", 22, "male", 0)` — notice you did **not** pass `self`; Python fills that in for you. You just hand over the four facts, in order.
- To **read a fact**, you write `owen.name`, `owen.age` — the **dot**, then the fact's name, with **no parentheses**. These are **attributes** — exactly the kind of thing you met as `ages.shape` in Session 4, except now *you* are the one who created them.

Because it's a blueprint, you can stamp out as many passengers as you like — each one separate, each remembering its own facts:

```python
laina = Passenger("Heikkinen, Miss. Laina", 26, "female", 1)

print(laina.name, "- age", laina.age, "- survived:", laina.survived)
```

One cutter, two cookies: `owen` and `laina` are two independent objects. Changing one never touches the other.

One last small thing — what if you just print the whole object?

```python
print(owen)
```

You get something cryptic, like **`<__main__.Passenger object at 0x…>`**. Don't be alarmed — that is just Python saying *"this is a Passenger object, living at this spot in the computer's memory."* It's ugly, but it confirms `owen` is a real, distinct thing. *(Making an object print itself **beautifully** is one of the nice touches you'll get almost for free next session.)*

***
## 5. Teaching the blueprint an *action* — a method

So far our passenger only **holds facts**. But an object's real power is that it can also **do things** — carry its own **actions**, called **methods**. A method is just a **function that lives inside the class** and works on `self` — *this* passenger.

Let's give `Passenger` the ability to answer a question: *"is this passenger a child?"* Update your class so it looks like this **in full** — the new `is_child` method goes **inside** the class, right under `__init__`:

```python
class Passenger:
    def __init__(self, name, age, sex, survived):
        self.name = name
        self.age = age
        self.sex = sex
        self.survived = survived

    def is_child(self):
        return self.age < 18
```

The new method reads almost like English: *"to check `is_child`, look at **this** passenger's own age (`self.age`) and report whether it is under 18."* It uses `self` to reach the passenger's **own** facts — the very facts `__init__` stored earlier.

Now stamp out a couple of passengers from the finished blueprint and **call the action** — with **parentheses**, because it's a thing the object *does*:

```python
owen = Passenger("Braund, Mr. Owen Harris", 22, "male", 0)
child = Passenger("Panula, Master. Eino Viljami", 1, "male", 0)

print("Is Owen a child?", owen.is_child())
print("Is Eino a child?", child.is_child())
```

`owen.is_child()` returns `False` (he's 22); `child.is_child()` returns `True` (Eino is 1). And there, side by side, is the whole distinction we've circled since Session 2, now fully in your hands:

- **`owen.age`** — a **fact** the object holds → an **attribute** → **no `()`**.
- **`owen.is_child()`** — an **action** the object performs → a **method** → **with `()`**.

You just wrote one of each.

***
## 6. Why bother? — and the one thing this can't do

Fair question: back in Session 2 we stored Owen in a **dictionary**. Why build a whole class?

Because an object gives you two things a loose dictionary can't:
- **A guaranteed shape.** *Every* `Passenger` has a name, an age, a sex, and a survived flag — always, in the same place. No passenger accidentally missing a field.
- **Built-in behaviour.** The data (`age`) and the actions that use it (`is_child`) travel **together**, in one tidy thing.

But here is the crack — and it matters. A plain class will cheerfully accept **complete nonsense**, without a word of complaint:

```python
ghost = Passenger("A. Ghost", "banana", "unknown", 5)

print("Age stored:", ghost.age)
```

Look at that: an **age of `"banana"`**, a **survived flag of `5`** — and Python just… stored them. No error, no warning. `print("Age stored:", ghost.age)` happily prints `banana`.

The trouble only strikes **later** — the moment something tries to do maths on that `"banana"` (say, averaging ages), your program crashes far away from where the bad data actually crept in. By then it's a detective hunt to find the culprit.

Wouldn't it be wonderful if the blueprint **checked** each fact *the instant* you created the object — *"an age must be a number; a survived flag must be 0 or 1"* — and refused the nonsense on the spot?

**That is exactly what the next session's tool, *pydantic*, does.** It lets you build a `Passenger`-style blueprint that **guarantees** its data is the right kind, catching mistakes at the door instead of deep inside your program. Everything you learned today — classes, attributes, `self` — is the foundation it's built on.

***
## What you can now do — and what's next

Today you finally met the "things" behind the dot, and built your own:

- **everything in Python is an object** — a bundle of **facts (attributes)** and **actions (methods)**;
- a **class** is a **blueprint**; an **object** is a thing **stamped out** from it;
- **`__init__`** is the set-up recipe, and **`self`** means *"this particular object"*;
- you **make** an object by calling the class (`Passenger(...)`), **read** its facts with a dot and no `()`, and **call** its actions with a dot and `()`;
- and a plain class **does not check its data** — the gap the finale closes.

> **⚠️ Quick reminders:**
> - **Class = blueprint, object = the thing built from it.** One class, many objects.
> - **`self`** is always the first thing in a method, and `self.x` is how an object reaches its own facts.
> - Watch your **colons** and your **indentation** — the two most common class errors.
> - **Attribute** = fact, **no `()`**. **Method** = action, **with `()`**. You can now write both.
