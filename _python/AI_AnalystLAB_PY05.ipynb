## 1. Why we need a *table*

Think of a **spreadsheet**, or a ship's old paper **passenger list**: one **row** per person, and a set of **columns** — name, age, ticket class, whether they survived. The quiet magic is that **each column holds a different *kind* of value**: the *name* column is text, the *age* column is numbers, the *survived* column is yes/no.

That is exactly the wall we hit last session. Remember our little story about *kinds of data*:

- a **list** (Session 2) happily **mixes** kinds — text and numbers together;
- a **NumPy array** (Session 4) is **strict** — every value must be the **same** kind.

A real dataset needs **both** at once: a *name* column of text **and** an *age* column of numbers, sitting side by side. Neither a plain list nor a single array can do that cleanly. The answer is a new kind of thing — a **table where each column keeps its own kind** — and that is precisely what **pandas** gives you. Its table is called a **DataFrame**.

> **pandas** (the name comes from *"Python Data Analysis"*) is another **add-on toolbox**, like NumPy — you installed it back in Session 0. By long tradition it is nicknamed **`pd`**.

## 4. The analyst's "first look" — four quick checks

Whenever an analyst opens a dataset for the first time, they run the **same four checks** before anything else. Make them a habit; you will see them at the start of nearly every session in the main course.

You already did check **1** — **`.head()`**, a peek at the first rows. Here are the other three. First, **how big is the table**, and **what kind is each column**:

```python
print("Size (rows, columns):", passengers.shape)   # a fact — no ()

print()
print("The kind of each column:")
print(passengers.dtypes)                            # a fact — no ()
```

- **`.shape`** reports **`(891, 12)`** — **891 rows** (passengers) and **12 columns** (facts about each one).
- **`.dtypes`** reports the **kind** held by each column — and here our whole *kinds of data* story finally pays off. Look: `Name` is **text**, `Age` is a **number** (`float64`), `Survived` is a **number** (`int64`) — **different kinds, one per column**, all in a single table. That is the thing neither a list nor an array could give us.

Notice there are **no parentheses** on `.shape` or `.dtypes`. That's the `( )`-test from Sessions 2 and 4: they are **facts the table simply has** (attributes), not **actions it performs** (methods). Compare `passengers.shape` (a fact, no `()`) with `passengers.head()` (an action — go fetch the first rows, so it takes `()`).

The fourth check, **`.describe()`**, gives a quick numeric summary of **every number column at once** — the descriptive statistics you met in Session 4 (count, mean, min, max, …), now for a whole table in one call:

```python
print(passengers.describe())
```

Each **column** of that summary is one of the number columns; each **row** is a statistic — the count, the average (`mean`), the smallest (`min`), the largest (`max`), and a few in between. In one glance you learn that the average fare was about 32, the oldest passenger was 80, and — reading the `mean` of the `Survived` column — that roughly **38%** of these passengers survived.

> **One honest note:** `.describe()` runs the numbers for **every** column that *looks* like a number, even ones where an average is meaningless — `PassengerId` is just a label, so its "average" tells you nothing. The tool can't know that; **you** do. Reading a summary always means knowing which columns actually deserve one.

***
## 5. Pulling out a single column

Most of the time you don't want the whole table — you want **one column**. You ask for it by its **name**, in square brackets — and if that feels familiar, it should: it is exactly like **looking up a key in a dictionary** from Session 2.

```python
ages = passengers["Age"]     # give me just the Age column
print(ages.head())
```

`passengers["Age"]` means *"give me the Age column"* — the very same move as `owen_dict["age"]` gave you one passenger's age in Session 2, only now you get the **whole column** for all 891 people.

And a single column behaves like a **labelled NumPy array**: every summary you learned in Session 4 works on it directly. For instance, the average age across the whole ship:

```python
print("Average age on board:", round(passengers["Age"].mean(), 1))
```

One line gives the average age of 891 people. *(You'll see roughly **29.7**. Hold that thought — it quietly skipped some passengers, and the next section explains why.)*

***
## 6. Keeping only the rows you want

Here is the move an analyst reaches for constantly: **keep only the rows that meet a condition**. It is the **boolean mask** from Session 4 — where you wrote `ages[ages >= 30]` to keep certain *values* — now applied to whole **rows of a table**.

You ask a yes/no question of a column, which gives a `True`/`False` for every row; put that inside square brackets, and pandas keeps only the rows where the answer is `True`. Let's keep only the passengers who **survived** (their `Survived` value is `1`):

```python
survivors = passengers[passengers["Survived"] == 1]
print("Number of survivors:", survivors.shape)
```

`survivors.shape` reports **`(342, 12)`** — of the 891 passengers, **342 survived**. Same stencil idea as Session 4, just keeping **rows** instead of single values.

Now combine the two tricks — filter the rows, then take a column's average — to ask a real question: *what was the average age of the survivors?*

```python
print("Survivors' average age:", round(survivors["Age"].mean(), 1))
```

Filtering rows and summarising a column, chained together, is the everyday bread-and-butter of data analysis — and you just did it in two short lines.

***
## 7. Real data is messy — missing values

Hand-typed data is always neat. **Real** data has **holes**. Glance back at the first rows from Section 3: the **`Cabin`** column showed **`NaN`** in places. `NaN` is pandas' way of writing **"there is no value here — it's missing."** Real datasets are full of them, and pretending otherwise is how analyses go wrong.

You can **count** the missing values in each column at a glance:

```python
print(passengers.isna().sum())
```

The count tells an honest story about this dataset: **177** passengers have **no recorded age**, **687** have no cabin listed, **2** have no port of departure — and the other columns are complete. That is completely normal for real-world data.

This also explains the little mystery from Section 5. When you asked for `passengers["Age"].mean()`, pandas quietly **skipped the 177 missing ages** and averaged the rest — which is why you still got a sensible **29.7**.

> **What we do — and don't do — today.** Right now we only **notice** and **count** missing values. Simply *recognising that they exist* is the first real skill of a data analyst. Deciding what to *do* about them — fill them in, or leave them out — is a proper topic of the **main AI Analyst LAB course**. For now, whenever you see **`NaN`**, read it as **"missing."**
