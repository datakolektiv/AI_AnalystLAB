## 1. `import` — borrowing a toolbox

Everything you have used so far — lists, dictionaries, `print`, `for`, `if`, numbers, text — is **part of the Python language itself**. It is there the instant Python starts; you never installed it, and you never have to ask for it. Think of the **controls built into every car** — the wheel, the pedals — always there, ready.

**NumPy is different: it is *not* part of Python.** It is an **add-on**, written by other people, which is exactly why you had to **install it separately** back in **Session 0** (that `pip install …` line you typed in the Terminal). It is like a **roof rack** you buy and bolt onto the car.

To use an add-on in a program you **`import`** it — that is the moment you "attach" the tool. By long tradition NumPy is given the short nickname **`np`**:

```python
import numpy as np

print("NumPy is ready.")
```

From now on, every NumPy tool starts with **`np.`** — that prefix simply means *"from the NumPy toolbox."*

Notice what you did **not** have to do: you never write `import list` or `import dict`, because those come with the language. You only ever `import` the add-ons you installed. *(One honest footnote: a few tools **do** come bundled with Python yet still need importing — you'll meet one or two — but the big data toolboxes like NumPy and pandas are genuine add-ons you installed yourself.)*

> **⚠️ If this line turns red** with **`ModuleNotFoundError: No module named 'numpy'`**, it means NumPy was never installed — the Session 0 step did not complete. Re-do the install from Session 0, or contact your instructor.

## 2. The array — a list, but all one kind

The star of NumPy is the **array**: a row of values, much like a list, but with two differences — every value must be the **same kind**, and NumPy can do maths on the whole thing at blazing speed.

We have carried our five passengers' ages since Session 2. Let's put them in an array — you make one by handing `np.array(...)` a list:

```python
ages_list = [22, 38, 26, 35, 35]   # our passengers' ages (a plain Python list)
ages = np.array(ages_list)         # turn the list into a NumPy array

print(ages)
```

It prints much like a list, just without the commas: `[22 38 26 35 35]`. You index and slice it exactly as you learned for lists — `ages[0]` is the first, `ages[-1]` the last. *(In real work you would often build that list straight from the passengers with the comprehension from Session 3: `[p["age"] for p in passengers]`.)*

***
## 3. The catch — an array holds only *one* kind

Here is the crucial difference from a list. Remember from Session 2 that a **list** happily mixes kinds — text, numbers, `True`/`False`, all together. A NumPy **array refuses.** Watch what happens when we try to mix a number and some text in one array:

```python
mixed = np.array([22, "male", True])
print(mixed)
print("The kind of thing inside:", mixed.dtype)
```

Look closely: the `22` and the `True` have **silently turned into text** — `['22' 'male' 'True']` — and `.dtype` (the *kind*) now reports something like `<U…`, NumPy's label for "text."

> **⚠️ This did *not* cause an error — which is exactly what makes it sneaky.** NumPy needs one single kind for the whole array, so when you mix kinds it quietly forces everything into the one kind that can hold them all: text. Your numbers stop being numbers, and maths on them will fail later. **The rule to remember: keep one kind of value per array** — usually, all numbers.

This strictness sounds like a pure limitation, and for a mixed table it is — but it is also the secret to NumPy's speed, and next session we'll see how **pandas** turns it into a strength.

***
## 4. The superpower — maths on the whole array at once

With an array of numbers you can do arithmetic on **every value at once**, with no loop at all. Where Session 3 needed a `for` loop to touch each number, NumPy touches them all in one stroke — like pressing the calculator button once for a whole column:

```python
print("Everyone one year older:", ages + 1)
print("Ages doubled:           ", ages * 2)
```

`ages + 1` added 1 to *each* age; `ages * 2` doubled *each* one — no loop written, and internally it runs far faster than a loop ever could. This "do it to the whole array at once" style is called **vectorised** maths, and it is why NumPy sits under every heavy calculation in this course.

## 5. Summaries — and functions over every value

Remember the loop you wrote in Session 3 to total the ages and divide for an average? NumPy gives you that — and more — in **one call each**. These are called **summaries**: they boil a whole array down to a single number.

You will meet a few of the everyday ones here. So the words are never a mystery, here is each in a single line:

- **Mean** (the *average*) — add up all the values and divide by how many there are.
- **Median** — the **middle** value once they are lined up smallest to largest (half the values fall below it, half above).
- **Minimum** and **maximum** — simply the smallest and the largest value.
- **Standard deviation** — how **spread out** the values are around the mean: a small number means they huddle close together, a large one means they are scattered.

> **These are the *descriptive statistics* the main course opens with.** Here we only see *how NumPy computes them* — one line each. Every one of them is explained **properly, from first principles**, in **Session 01 of the AI Analyst LAB course** that follows this crash course. For now, just get comfortable with the names and the one-line ideas.

Let's compute them all on our passengers' ages:

```python
print("Average (mean):         ", ages.mean())
print("Middle value (median):  ", np.median(ages))
print("Total:                  ", ages.sum())
print("Youngest:               ", ages.min())
print("Oldest:                 ", ages.max())
print("Spread (std. deviation):", round(ages.std(), 2))
```

One line for the average, instead of a loop and a running total. Two things worth noticing:

- The **median** — the middle value once the ages are sorted — is written **`np.median(ages)`**, a NumPy *function*, because an array happens not to carry a `.median()` *method* the way it carries `.mean()`. Many summaries come in **both** styles: the method `ages.mean()` and the function `np.mean(ages)` give the very same answer. When a method doesn't exist, reach for the `np.` function.
- The "spread" number — the *standard deviation* — is one you'll meet properly in the statistics reviews.

#### Functions that touch *every* value

The summaries above each squeezed a whole array into **one** number. NumPy also has functions that do the opposite: they apply the same operation to **every** value and hand back a whole **new array**. The classic is the square root, `np.sqrt`:

```python
print("Square root of each age:", np.round(np.sqrt(ages), 1))   # np.round(..., 1) just keeps it tidy
```

`np.sqrt(ages)` took the square root of *every* value at once and gave back a new array of five results — the same *"do it to the whole array"* idea as `ages * 2`, only through a named function. (We wrapped it in `np.round(..., 1)` purely so the decimals stay readable.) A few you will meet, all working value by value:

- `np.sqrt(arr)` — the square root of each value
- `np.round(arr, 1)` — round each value (here, to 1 decimal)
- `np.abs(arr)` — the size of each value, ignoring its sign

*(And in the statistics reviews, `np.log2` and `np.exp` turn up the same way — one result per value — for entropy and probability.)*

The distinction to keep in mind: a **summary** like `ages.mean()` hands back **one number**; an **element-wise function** like `np.sqrt(ages)` hands back a **whole array** — one result per value.

## 6. Your first real *attributes* — `.shape`, `.size`, `.dtype`

Back in Session 2 we drew a line between a **method** — an *action* a thing does, written with `( )` — and an **attribute** — a *fact* a thing simply has, written with **no** `( )`. We promised you would meet a real attribute soon. Here it is.

`ages.mean()` is a **method**: it has `( )` because it *does* something (goes and computes the average). But an array also carries plain **facts about itself** that you read with **no parentheses**:

```python
print("How many values (size):", ages.size)     # a fact — no ()
print("Shape:                 ", ages.shape)     # a fact — no ()
print("Kind of value (dtype): ", ages.dtype)     # a fact — no ()
```

No parentheses on any of those — because `size`, `shape`, and `dtype` are **facts the array has**, not actions it does. Compare the two side by side:

- `ages.size` → a *fact*: how many values. **No `( )`.** → an **attribute**
- `ages.mean()` → an *action*: go and average them. **With `( )`.** → a **method**

That is the `( )`-test from Session 2, now concrete. And `.dtype` is the very thing that caught us out in §3 — it names the single kind an array holds (`int64`, NumPy's name for whole numbers, here).

> **⚠️ Don't cross the two up.** Writing `ages.shape()` (with parentheses, as if it were an action) gives a red **`TypeError`** — a fact is not something you "call." And `ages.mean` *without* parentheses computes nothing — it just prints the method itself, like the `owen_dict.keys` surprise from Session 2.

## 7. Picking values by a condition — boolean masks

One more NumPy trick you will lean on constantly. Write a **comparison** against a whole array, and you get back an array of `True`/`False` — one answer per value:

```python
print(ages >= 30)     # is each passenger 30 or older?
```

That array of `True`/`False` is called a **mask** — picture a stencil laid over your data. Put it inside square brackets, and NumPy keeps only the values where the mask is `True`:

```python
older = ages[ages >= 30]
print("Ages of passengers 30 or older:", older)
```

In a single line, you filtered the data down to just the passengers aged 30 or more. Hold on to this idea — **selecting the items that meet a condition** is one of the most common things an analyst ever does, and in **Session 5** you will do exactly this to whole **rows** of a real table (*"show me only the survivors"*).

## 8. A quick peek — random numbers

Much of the AI Analyst LAB course *simulates* things — flipping coins ten thousand times, drawing random samples to watch a bell curve appear. NumPy is what makes that possible. You start a **random-number generator** (with a fixed `seed`, so the results are repeatable), then ask it for numbers:

```python
rng = np.random.default_rng(seed=42)     # a repeatable random-number generator

print("Five random whole numbers, 1 to 100:", rng.integers(1, 100, size=5))
```

Because we fixed the `seed`, you get the **same** five "random" numbers every time you run it — which is exactly what we want for results other people can reproduce. This is just a taste; the statistics reviews use this engine heavily. For now, simply know it is there.

## Where NumPy stops, and pandas begins

NumPy is superb at **one column of numbers**, crunched fast. But a real dataset is a whole **table**, and that is where NumPy runs out of road — and where **pandas** (Session 5) takes over:

- **Mixed columns.** Our passengers have a *text* name, a *number* age, and a *True/False* survived — side by side. A NumPy array can hold only **one kind for the whole thing** (§3). **pandas gives every column its own kind** — the "smart mixing" we promised back in Session 2.
- **Names, not positions.** With NumPy you'd fetch age by *position* (which column was that again?). **pandas lets you ask by name** — `table["age"]` — just like looking up a dictionary label.
- **Real files, gaps and all.** **pandas loads a whole CSV in one line** — all **891** real passengers — and copes with **missing values** (the real Age column has gaps NumPy has no tidy answer for).
- **Group-and-summarise.** *"Survival rate by sex"* in a single line — a verb NumPy simply does not have.

So think of it this way: **NumPy is the fast engine; pandas is the whole vehicle built around it.**
