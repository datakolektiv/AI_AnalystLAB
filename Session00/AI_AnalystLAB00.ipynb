#### 0.4 Get the course code from GitHub

The course code lives in a **private** GitHub repository: [`datakolektiv/AI_AnalystLAB`](https://github.com/datakolektiv/AI_AnalystLAB). To download it onto your computer ("clone" it, in Git terminology), you need three things:

1. A free **GitHub account**.
2. **Access** to the repository — your instructor will invite you by email.
3. A **Personal Access Token (PAT)** — a long string that lets Git authenticate as you. We'll create one in this section.

Follow the four sub-steps below in order. Total time: about **5 minutes**.

---

##### 0.4.1 Create a GitHub account (skip if you already have one)

1. Go to [https://github.com/signup](https://github.com/signup) and create a free account.
2. Confirm your email address (GitHub will send you a verification email — check your inbox).
3. **Tell your instructor your GitHub username** (the short name you chose, *not* your email) so they know who to invite.

---

##### 0.4.2 Accept the invitation to the course repository

Your instructor will invite you to the `datakolektiv/AI_AnalystLAB` repository. Once they do:

1. Check the inbox of the email associated with your GitHub account. You will receive a message from `noreply@github.com`. The subject line looks like **"\[GitHub\] @goranmilovanovic has invited you to..."**.

3. If you are not signed in to GitHub, sign in.

> **Verification.** Open [https://github.com/datakolektiv/AI_AnalystLAB](https://github.com/datakolektiv/AI_AnalystLAB) in your browser while signed in. If the page loads and shows the repository, you have access. If you see "404 — page not found", the invitation has not been accepted yet — go back to step 2, or contact your instructor.

---

##### 0.4.3 Generate a Personal Access Token (PAT)

A Personal Access Token is like a password that Git uses on your behalf. We will create a *fine-grained* token that is restricted to **only this one repository** and **only the read permission** — so it is safe to use.

1. Open this link in your browser: [https://github.com/settings/personal-access-tokens/new](https://github.com/settings/personal-access-tokens/new)
2. If GitHub asks you to confirm your identity (a dialog called "Sudo mode"), enter your password.
3. Fill in the form **exactly** as follows:
   - **Token name:** `AI_AnalystLAB-clone` (any descriptive name will do)
   - **Expiration:** select **90 days** (or longer if you prefer)
   - **Description:** `For cloning the AI Analyst LAB course repo` (optional)
   - **Resource owner:** click the dropdown and select **`datakolektiv`**
   - **Repository access:** click **"Only select repositories"**, then in the new dropdown that appears choose **`datakolektiv/AI_AnalystLAB`**
   - **Permissions** → expand the **"Repository permissions"** section → scroll down to **"Contents"** → set it to **"Read-only"**
4. Click the green **"Generate token"** button at the bottom of the page.
5. **A new page appears with your token.** It is a long string starting with `github_pat_…`.
   - **Copy it immediately** using the small "copy" button next to the token.
   - **Paste it somewhere safe** — for example, into a temporary text file on your desktop, or into a password manager. **GitHub will not show this token again.** If you lose it, you will have to generate a new one.

---

##### 0.4.4 Clone the repository

Now we use the PAT to clone the repository onto your computer.

1. Open your terminal:
   - **Windows:** open PowerShell.
   - **macOS / Linux:** open Terminal.
2. Navigate to your course directory. Use the line that matches your operating system:

   **Windows:**
   ```powershell
   cd C:\Users\<your-name>\ai_analyst_lab
   ```

   **macOS / Linux:**
   ```bash
   cd ~/ai_analyst_lab
   ```

3. Run the clone command (the same on every operating system):
   ```bash
   git clone https://github.com/datakolektiv/AI_AnalystLAB.git
   ```
4. Git will prompt you for credentials:
   - **`Username for 'https://github.com':`** — type your **GitHub username** (the one you chose when signing up — *not* your email address) and press ENTER.
   - **`Password for 'https://<your-username>@github.com':`** — **paste the PAT** you copied in 0.4.3 and press ENTER.

   > ⚠️ **Important:** when you paste the PAT, the terminal **does not show the characters** — not even dots. That is a security feature, not a bug. Just paste and press ENTER.
5. After a few seconds, Git prints something like:
   ```
   Cloning into 'AI_AnalystLAB'...
   remote: Enumerating objects: ...
   Receiving objects: 100% ...
   ```
   When the prompt returns, the clone is finished. Run `ls` (or `dir` on Windows) and confirm that a new **`AI_AnalystLAB`** folder appeared inside your course directory.

---

##### 0.4.5 Pulling future updates

Your PAT is now cached on your computer, so Git will not ask for it again. Whenever your instructor adds new course material, run:

```bash
cd AI_AnalystLAB
git pull
```

You will get any new files or updates to existing ones.

> The PAT will expire after 90 days (or whatever you chose). When it expires, you will be asked for credentials again — repeat **0.4.3** to generate a fresh token.

##### Common things that go wrong

- **"Authentication failed"** — most often, this means you typed your *email address* as the username instead of your GitHub *username*, or the token was pasted with extra whitespace. Try the clone again carefully.
- **"Repository not found" or 404** — you have not accepted the invitation. Go back to **0.4.2**.
- **A browser window opened asking you to sign in to GitHub** — that is fine, sign in there and let Git Credential Manager finish. The clone should then proceed automatically without asking for the PAT. If that works, you do not strictly need the PAT — but keep it as a backup.

### 1. Smoke test: Simple Linear Regression with `scikit-learn`

If your installation went well, the cells below will run from top to bottom without errors. We use the classic **Iris** dataset to fit a tiny linear regression model and draw a chart — just enough to confirm that NumPy, Pandas, scikit-learn, Matplotlib, and Seaborn are all working together.

Import the packages we need.

````python
# - packages
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
````

Load the Iris dataset that ships with `scikit-learn`.

````python
# - data
from sklearn.datasets import load_iris
ds = load_iris()
data_set = pd.DataFrame(data=ds.data, columns=ds.feature_names)
data_set.head(10)
````

Inspect the shape of the data — Iris has 150 rows and 4 columns.

````python
data_set.shape
````

Compute pairwise correlations between the four numeric columns.

````python
data_set.corr(numeric_only=True)
````

Visualize the correlation matrix as a heatmap with **Seaborn**.

````python
# - Seaborn
import seaborn as sns
cor_mat = data_set.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(cor_mat, annot=True)
plt.show()
````

Confirm the data types of every column.

````python
data_set.dtypes
````

Now fit a **simple linear regression** with `scikit-learn`: predict `petal length (cm)` from `sepal length (cm)`.

````python
# - numpy + scikit-learn
import numpy as np
from sklearn.linear_model import LinearRegression

# - reshape because we have a single predictor
X = np.array(data_set['sepal length (cm)']).reshape(-1, 1)
y = np.array(data_set['petal length (cm)']).reshape(-1, 1)

# - fit
linear_model = LinearRegression().fit(X, y)

# - report
print("Slope of the regression line:    " + str(linear_model.coef_.item()))
print("Intercept of the regression line: " + str(linear_model.intercept_.item()))
````

Visualize the regression line on top of the data.

````python
plt.figure(figsize=(8, 8))
plt.box(False)
plt.scatter(X, y, color='green', marker='.')
plt.plot(X, linear_model.predict(X), color='blue', linewidth=0.8)
plt.title('Iris: sepal length vs. petal length')
plt.xlabel('sepal length (cm)')
plt.ylabel('petal length (cm)')
plt.show()
````
